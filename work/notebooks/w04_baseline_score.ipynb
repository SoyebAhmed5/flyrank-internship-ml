### My baseline rule

I will build a simple, rule-based score to prioritize content items for human review in the Refresh / Content Opportunity Scoring lane.

Before choosing the final rule, I will inspect two signals: content age (staleness) and search impressions (volume). I will use bucket tables with sample counts to check whether these signals show a useful relationship with the observed decline label.

The final score will use only fields available at the decision moment. It will not use the decline label, label-derived fields, existing product flags, or future-window information. Each item will receive one score, one reason code, and one action label.

The ranking is a decision-support tool, not proof that a content item needs refreshing or that a refresh will improve its performance.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.



# Inspect the actual columns first
print("Columns:", df.columns.tolist())

# Create the starter proxy label if it is not already present
if "is_declining_label" not in df.columns:
    if "trend_direction" not in df.columns:
        raise ValueError(
            "Neither is_declining_label nor trend_direction exists. "
            "Check the printed column names before continuing."
        )

    df["is_declining_label"] = (
        df["trend_direction"].astype(str).str.lower() == "down"
    ).astype(int)

required = [
    "content_age_days",
    "impressions_90d",
    "is_declining_label",
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

print("\nSignal-check columns are ready.")
print("Rows:", len(df))
print("Declining-label rate:", round(df["is_declining_label"].mean(), 3))


```
