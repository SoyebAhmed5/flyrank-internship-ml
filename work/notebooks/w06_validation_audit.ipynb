### Overall Methodology Assessment

The research uses aggregated Google Search Console and Google Analytics data to study content performance.

The paper reports direct aggregate comparisons as its main evidence and describes its machine learning experiments as exploratory.

The machine learning appendix mentions 80/20 train-test splits, but it does not clearly specify whether the splits were grouped by client or separated by time.

The paper also does not report p-values or confidence intervals for its findings.

**My assessment:**

The paper provides useful descriptive insights about content performance, but its findings should not be interpreted as proof of causation.

A stronger validation approach would include:
- Grouped validation to prevent the same client's content from appearing in both training and testing sets.
- Time-based validation when predicting future performance.
- Confidence intervals and statistical significance testing.
- Controls for possible confounding factors.
- Clear documentation of feature timing and label construction.

### Validation Approach

For this experiment, I use the Week-5 Logistic Regression model with the same target, features, preprocessing steps, and model configuration.

I compare two validation strategies:

1. Random-row split: Rows are randomly divided into training and testing sets. Content from the same client may appear in both sets.
2. Client-grouped split: All content belonging to a client is assigned entirely to either training or testing.

The random-row split is included as a diagnostic comparison. The client-grouped split is the primary evaluation because it measures how well the model generalizes to clients it has not seen during training.

Both experiments use an 80/20 train-test ratio.

The starter dataset does not contain an explicit reporting date that allows a reliable chronological split. Therefore, this experiment evaluates generalization to unseen clients, not future time periods.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

from sklearn.model_selection import train_test_split, GroupShuffleSplit
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, roc_auc_score

# Prepare the dataset
model_df = df.dropna(
    subset=["trend_direction", "client_id"]
).copy()

# Create the target
model_df["target_declining"] = (
    model_df["trend_direction"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("down")
    .astype(int)
)

# Features
numeric_features = [
    "content_age_days",
    "search_volume",
    "competition",
    "cpc",
    "word_count",
    "char_count"
]

categorical_features = [
    "content_type",
    "main_intent",
    "competition_level"
]

feature_columns = numeric_features + categorical_features

X = model_df[feature_columns]
y = model_df["target_declining"]
groups = model_df["client_id"]


# Build the same model for both experiments
def build_model():

    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ])

    return Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ])


# --------------------------------
# 1. Random-row split
# --------------------------------

X_train_random, X_test_random, y_train_random, y_test_random = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)

random_model = build_model()

random_model.fit(X_train_random, y_train_random)

random_prob = random_model.predict_proba(
    X_test_random
)[:, 1]


# --------------------------------
# 2. Client-grouped split
# --------------------------------

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train_group = X.iloc[train_idx]
X_test_group = X.iloc[test_idx]

y_train_group = y.iloc[train_idx]
y_test_group = y.iloc[test_idx]

group_model = build_model()

group_model.fit(X_train_group, y_train_group)

group_prob = group_model.predict_proba(
    X_test_group
)[:, 1]


# --------------------------------
# 3. Compare results
# --------------------------------

random_overlap = (
    set(groups.loc[X_train_random.index])
    & set(groups.loc[X_test_random.index])
)

group_overlap = (
    set(groups.iloc[train_idx])
    & set(groups.iloc[test_idx])
)

results = pd.DataFrame([
    {
        "Split": "Random Row",
        "Train Rows": len(X_train_random),
        "Test Rows": len(X_test_random),
        "Test Base Rate": y_test_random.mean(),
        "Average Precision": average_precision_score(
            y_test_random, random_prob
        ),
        "ROC-AUC": roc_auc_score(
            y_test_random, random_prob
        ),
        "Client Overlap": len(random_overlap)
    },
    {
        "Split": "Client Grouped",
        "Train Rows": len(X_train_group),
        "Test Rows": len(X_test_group),
        "Test Base Rate": y_test_group.mean(),
        "Average Precision": average_precision_score(
            y_test_group, group_prob
        ),
        "ROC-AUC": roc_auc_score(
            y_test_group, group_prob
        ),
        "Client Overlap": len(group_overlap)
    }
])

display(results.round(4))

print("Overall target base rate:", round(y.mean(), 4))


forbidden_features = [
    "trend_direction",
    "trend_pct",
    "client_id",
    "content_id",
    "impressions_last_30d",
    "impressions_prev_30d",
    "clicks_last_30d",
    "impressions_90d",
    "ctr",
    "avg_position"
]

included_forbidden = (
    set(feature_columns) & set(forbidden_features)
)

print("Final feature list:")
print(feature_columns)

print("\nNumber of features:", len(feature_columns))

print("\nForbidden features found:", included_forbidden)

assert len(included_forbidden) == 0

print("\nLeakage feature checks passed!")
```
