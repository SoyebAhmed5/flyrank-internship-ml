## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

### Initial observations

The dataset contains **[number of rows] rows and [number of columns] columns**, representing **[number of unique content items] unique content items** across **[number of unique clients] clients**. I observed that **[number of downward-trend records] records ([percentage]%)** have a downward trend, and the median number of impressions over 90 days is **[median impressions]**.

These initial measurements suggest that the dataset provides information that may help investigate content-review opportunities. However, a downward trend alone does not establish that a page needs updating or that an update would improve its performance. Further analysis is needed to determine whether the available signals can support useful prioritization.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "SoyebAhmed5/flyrank-internship-ml/main/"
    "data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(url)

print("Dataset dimensions:", df.shape)
print("Unique content items:", df["content_id"].nunique())
print("Unique clients:", df["client_id"].nunique())

required_columns = [
    "content_id",
    "client_id",
    "trend_direction",
    "impressions_90d",
    "content_age_days",
]

missing = [c for c in required_columns if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

down_count = int(df["trend_direction"].eq("down").sum())
down_pct = df["trend_direction"].eq("down").mean() * 100

print(f"Records with downward trend: {down_count:,}")
print(f"Downward-trend percentage: {down_pct:.2f}%")
print(
    "Median 90-day impressions:",
    round(df["impressions_90d"].median(), 2),
)
```
